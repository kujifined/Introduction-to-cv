# Steel surface defect classification · live demo

Four frozen seed-43 checkpoints: ResNet-18 and EfficientNet-B0, each with minimal and camera-aware training. All predictions are computed on the CPU when you run a scene. See [the setup guide](README_LIVE_DEMO.md).

**Before the talk:** run the preparation cell once. Then run scenes 1–3 individually.

```python
import sys
from pathlib import Path

import torch
from IPython.display import HTML, display

cwd = Path.cwd()
demo_dir = next(
    (
        path.resolve()
        for path in (cwd, cwd / "notebooks", cwd / "project" / "notebooks")
        if (path / "demo_assets" / "demo_cases.json").is_file()
    ),
    None,
)
if demo_dir is None:
    raise FileNotFoundError("Open the notebook from its directory or the project root.")
sys.path.insert(0, str(demo_dir))

from live_demo import LiveDemo, checkpoint_source, render_case  # noqa: E402

torch.set_num_threads(min(4, torch.get_num_threads()))
weights_source = checkpoint_source(demo_dir)
demo = LiveDemo(demo_dir / "demo_assets", weights_source)
demo_results = {}
_ = demo.predict_case("clean_reference")  # Warm up before the timed presentation.
print("Ready: four verified checkpoints, seed 43, CPU. Run scenes 1–3 below.")
```

## 1 · Clean crazing

The four models receive the same clean crop.

```python
demo_results["clean_reference"] = demo.predict_case("clean_reference")
display(HTML(render_case(demo_results["clean_reference"])))
```

## 2 · The same crop, blurred

Apply the fixed Gaussian blur (severity 3) to the original crop before resizing.

```python
demo_results["same_image_blurred"] = demo.predict_case("same_image_blurred")
display(HTML(render_case(demo_results["same_image_blurred"])))
```

## 3 · A real clean-test error

Compare the two most likely classes for this scratch.

```python
demo_results["observed_failure"] = demo.predict_case("observed_failure")
display(HTML(render_case(demo_results["observed_failure"])))
```
